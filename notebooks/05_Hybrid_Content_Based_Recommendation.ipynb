```python
import ast

def get_collection_name(x):
  if isinstance(x, float):
    return ""
  try:
    data_collection = ast.literal_eval(x)
    return data_collection['name']
  except (ValueError, KeyError, TypeError, NameError, SyntaxError):
    return ""
```

```python
def get_extraction_name(x):
  if isinstance(x, float):
    return ""
  try:
    data_list = ast.literal_eval(x)
    return [item['name'] for item in data_list]
  except (ValueError, KeyError, TypeError, NameError, SyntaxError):
    return ""
```

```python
kolom_semistructerd = ['genres', 'production_companies']
for col in kolom_semistructerd:
  df_select[col] = df_select[col].apply(get_extraction_name)
```

```python
df_select['belongs_to_collection'] = df_select['belongs_to_collection'].apply(get_collection_name)
```

**Preprocessing**

```python
import numpy as np
from scipy.sparse import csr_matrix
from sklearn.feature_extraction.text import TfidfVectorizer

def build_tfidf_matrix(series, max_features=5000):
  """
  Mengubah kolom teks menjadi matrikx TF-IDF
  """
  if not series.str.len().gt(0).any():
    return None, csr_matrix((len(series), 1), dtype=np.float)

  vectorizer = TfidfVectorizer(
      stop_words='english',
      max_features=max_features,
      ngram_range=(1,2)
  )
  matrix = vectorizer.fit_transform(series)
  return vectorizer, matrix
```

```python
overview_vectorizer, overview_matrix = build_tfidf_matrix(df_select['overview'])
tagline_vectorizer, tagline_matrix = build_tfidf_matrix(df_select['tagline'])
print("Overview:", overview_matrix.shape)
print("Tagline:", tagline_matrix.shape)
```

```python
# Fungsi mengubah list menjadi himpunan saat menghitung Jaccard similarity
def to_name_set(value):
  """
  Menormalkan list nama menjadi set nama unik.
  Mendukung list, tuple, set dan nilai kosong
  """
  if not isinstance(value, (list, tuple, set)):
    return set()
  return {
      str(item).strip().casefold()
      for item in value
      if item is not None and str(item).strip()
  }
```

```python
genre_sets = df_select['genres'].apply(to_name_set)
company_sets = df_select['production_companies'].apply(to_name_set)
```

```python
"""
Dua film dianggap memiliki koleksi yang sama hanya
jika keduanya mempunyai nilai koleksi yang terisi dan nilainya sama
"""
def normalize_collection(value):
  if value is None:
    return None

  if pd.isna(value):
    return None

  value = str(value).strip()
  if not value:
    return None
  return value.casefold()
```

```python
collection_values = (df_select['belongs_to_collection'].apply(normalize_collection))
```

```python
# Membangun fungsi similarity
def jaccard_similarity(set_a, set_b):
  # Jaccard similarity untuk dua set.
  if not set_a or not set_b:
    return 0.0
  return len(set_a & set_b) / len(set_a | set_b)
```

```python
def collection_similarity(collection_a, collection_b):
  # Jika kedua film memiliki koleksi yang sama.
  if collection_a is None or collection_b is None:
    return 0.0
  return float(collection_a == collection_b)
```

```python
def text_similarity(query_idx, matrix):
  #Cosine similarity antara film query dan semua film
  return cosine_similarity(
      matrix[query_idx],
      matrix
  ).ravel()
```

```python
# Bobot
Weights = {
    'overview':0.35,
    'tagline':0.10,
    'genres': 0.25,
    'collection': 0.10,
    'companies': 0.20
}
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from IPython.display import display

#Fungsi rekomendasi hybrid
def recommend_hybrid(movie_title, top_n=10):
  # Cari judul film yang cocok tanpa membedakan huruf besar/kecil
  matches = df_select.index[df_select['title'].fillna('').str.casefold() == str(movie_title).casefold()].tolist()

  if not matches:
    raise ValueError(f"Film '{movie_title}' tidak ditemukan")

  query_idx = matches[0]

  # Skor berbasis teks
  overview_scores = text_similarity(query_idx, overview_matrix)
  tagline_scores = text_similarity(query_idx, tagline_matrix)

  # Skor berbasis metadata
  query_genres = genre_sets[query_idx]
  query_companies = company_sets[query_idx]
  query_collection = collection_values.iloc[query_idx]

  genre_scores = np.array([
      jaccard_similarity(query_genres, candidate)
      for candidate in genre_sets
  ])

  company_scores = np.array([
      jaccard_similarity(query_companies, candidate)
      for candidate in company_sets
  ])

  collection_scores = np.array([
      collection_similarity(query_collection, candidate)
      for candidate in collection_values
  ])

  # Gabungkan seluruh sinyal similarity
  hybrid_scores = (
      Weights['overview'] * overview_scores +
      Weights['tagline'] * tagline_scores +
      Weights['genres'] * genre_scores +
      Weights['collection'] * collection_scores +
      Weights['companies'] * company_scores
  )

  # Jangan rekomendasikan film itu sendiri
  hybrid_scores[query_idx] = -np.inf

  # Urutkan skor tertinggi
  ranked_idx = np.argsort(hybrid_scores)[::-1]
  ranked_idx = ranked_idx[:top_n]
  results = df_select.iloc[ranked_idx][
      ['title','genres','overview']
  ].copy()
  results.insert(
      1,
      'hybrid_score',
      hybrid_scores[ranked_idx]
  )

  results = results.reset_index(drop=True)
  return results
```

```python
# Uji rekomendasi
recommendations = recommend_hybrid(
    movie_title='The Dark Knight',
    top_n = 10
)
display(recommendations)
```
